```python
import pandas as pd
```

```python
costo_2024 = pd.read_excel("..\data\COSTOS DE MATERIALES 2024.xlsx")
costo_2025 = pd.read_excel("..\data\COSTOS DE MATERIALES 2025.xlsx")
ingresos = pd.read_excel("..\data\INSUMOS OPERACIONALES STOCK RELIX.xlsx", sheet_name='INGRESO', skiprows=9)
```

```python
costo_2024["mes_año"] = costo_2024["Mes"].astype("Int64").astype(str).str.zfill(2) + "/2024"
costo_2024 = costo_2024.drop(columns=["Mes"])
costo_2025["mes_año"] = costo_2025["Mes"].astype("Int64").astype(str).str.zfill(2) + "/2025"
costo_2025 = costo_2025.drop(columns=["Mes"])
```

```python
costo_total = pd.concat([costo_2024, costo_2025], ignore_index=True)
costo_total
```

```python
costo_total = costo_total.dropna(how="all").reset_index(drop=True)
costo_total
```

```python
costo_total = costo_total.dropna(subset=['Centro de coste'])
costo_total
```

```python
costo_total['mes_año'] = pd.to_datetime(costo_total['mes_año'], format="%m/%Y")
```

```python
import altair as alt

# Crear columna de fecha para ordenar correctamente
costo_plot = costo_total.copy()
costo_plot['fecha'] = pd.to_datetime(costo_plot['mes_año'], format='%m/%Y', errors='coerce')
costo_plot = costo_plot.dropna(subset=['fecha', 'Denom.clase de coste'])

# Agrupar por mes y categoría
costo_mensual = (
    costo_plot
    .groupby([pd.Grouper(key='fecha', freq='MS'), 'Denom.clase de coste'])['Valor/Moneda objeto']
    .sum()
    .reset_index()
)
costo_mensual.rename(columns={
    'Valor/Moneda objeto': 'Costo Total (USD)',
    'Denom.clase de coste': 'Categoría'
}, inplace=True)

# Gráfico interactivo con Altair
chart = alt.Chart(costo_mensual).mark_bar().encode(
    x=alt.X('yearmonth(fecha):T', title='Mes', axis=alt.Axis(format='%b %Y', labelAngle=-45)),
    y=alt.Y('Costo Total (USD):Q', title='Costo Total (USD)', axis=alt.Axis(format='$,.0f')),
    color=alt.Color('Categoría:N', title='Categoría'),
    xOffset='Categoría:N',
    tooltip=[
        alt.Tooltip('yearmonth(fecha):T', title='Mes', format='%b %Y'),
        alt.Tooltip('Categoría:N'),
        alt.Tooltip('Costo Total (USD):Q', format='$,.2f')
    ]
).properties(
    width=900,
    height=450,
    title='Costo Mensual por Categoría (2024 - 2025)'
).interactive()

chart
```

```python
import altair as alt

# --- Costos de materiales (SAP) ---
costo_plot2 = costo_total.copy()
costo_plot2['fecha'] = pd.to_datetime(costo_plot2['mes_año'], format='%m/%Y', errors='coerce')
costo_plot2 = costo_plot2[costo_plot2['fecha'] >= '2025-10-01']
costo_mensual_sap = (
    costo_plot2.groupby(pd.Grouper(key='fecha', freq='MS'))['Valor/Moneda objeto']
    .sum().reset_index()
)
costo_mensual_sap.rename(columns={'Valor/Moneda objeto': 'Costo'}, inplace=True)
costo_mensual_sap['Costo'] = costo_mensual_sap['Costo'] * 910
costo_mensual_sap['Acumulado'] = costo_mensual_sap['Costo'].cumsum()
costo_mensual_sap['Fuente'] = 'Costos Materiales (SAP)'

# --- Ingresos a bodega ---
ingresos_plot2 = ingresos.copy()
ingresos_plot2['fecha'] = pd.to_datetime(ingresos_plot2['FECHA DE INGRESO'], errors='coerce')
ingresos_plot2 = ingresos_plot2[ingresos_plot2['fecha'] >= '2025-10-01']
ingreso_mensual2 = (
    ingresos_plot2.groupby(pd.Grouper(key='fecha', freq='MS'))['TOTAL NETO']
    .sum().reset_index()
)
ingreso_mensual2.rename(columns={'TOTAL NETO': 'Costo'}, inplace=True)
ingreso_mensual2['Acumulado'] = ingreso_mensual2['Costo'].cumsum()
ingreso_mensual2['Fuente'] = 'Ingresos Bodega'

# --- Combinar ---
comparacion = pd.concat([costo_mensual_sap, ingreso_mensual2], ignore_index=True)

# --- Gráfico acumulado ---
line = alt.Chart(comparacion).mark_line(point=True, strokeWidth=3).encode(
    x=alt.X('yearmonth(fecha):T', title='Mes', axis=alt.Axis(format='%b %Y', labelAngle=-45)),
    y=alt.Y('Acumulado:Q', title='Costo Acumulado (USD)', axis=alt.Axis(format='$,.0f')),
    color=alt.Color('Fuente:N', title='Fuente'),
    tooltip=[
        alt.Tooltip('yearmonth(fecha):T', title='Mes', format='%b %Y'),
        alt.Tooltip('Fuente:N'),
        alt.Tooltip('Costo:Q', title='Costo Mensual', format='$,.2f'),
        alt.Tooltip('Acumulado:Q', title='Acumulado', format='$,.2f')
    ]
).properties(
    width=800,
    height=450,
    title='Costo Total Acumulado desde Octubre 2025: Costos SAP vs Ingresos Bodega'
).interactive()

line
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

# Preparar datos
costo_ts = costo_total.copy()
costo_ts['fecha'] = pd.to_datetime(costo_ts['mes_año'], format='%m/%Y', errors='coerce')
costo_ts = costo_ts.dropna(subset=['fecha'])
costo_ts['Año'] = costo_ts['fecha'].dt.year
costo_ts['Mes'] = costo_ts['fecha'].dt.month

# Agrupar por mes y año
costo_mensual = (
    costo_ts
    .groupby(['Año', 'Mes'])['Valor/Moneda objeto']
    .sum()
    .reset_index()
    .sort_values(['Año', 'Mes'])
)
costo_mensual['Acumulado'] = costo_mensual.groupby('Año')['Valor/Moneda objeto'].cumsum()

# Gráfico
fig, ax = plt.subplots(figsize=(12, 5))

for año, grupo in costo_mensual.groupby('Año'):
    ax.plot(grupo['Mes'], grupo['Acumulado'], marker='o', linewidth=2.5, label=str(año))

ax.set_xlabel('Mes')
ax.set_ylabel('Costo Acumulado (USD)')
ax.set_title('Costos Acumulados por Año')
ax.set_xticks(range(1, 13))
ax.set_xticklabels(['Ene','Feb','Mar','Abr','May','Jun','Jul','Ago','Sep','Oct','Nov','Dic'])
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
ax.legend(title='Año')
ax.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# dejar los stocks negativos en 0
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np

# Preparar datos
costo_ts = costo_total.copy()
costo_ts['fecha'] = pd.to_datetime(costo_ts['mes_año'], format='%m/%Y', errors='coerce')
costo_ts = costo_ts.dropna(subset=['fecha'])
costo_ts['Año'] = costo_ts['fecha'].dt.year
costo_ts['Mes'] = costo_ts['fecha'].dt.month

# Agrupar por mes y año
costo_mensual = (
    costo_ts
    .groupby(['Año', 'Mes'])['Valor/Moneda objeto']
    .sum()
    .reset_index()
    .sort_values(['Año', 'Mes'])
)
costo_mensual['Acumulado'] = costo_mensual.groupby('Año')['Valor/Moneda objeto'].cumsum()

# Presupuestos anuales
presupuestos = {2024: 5_300_000, 2025: 5_800_000}
meses_labels = ['Ene','Feb','Mar','Abr','May','Jun','Jul','Ago','Sep','Oct','Nov','Dic']

fig, axes = plt.subplots(1, 2, figsize=(18, 6), sharey=False)

for idx, (año, presupuesto_anual) in enumerate(presupuestos.items()):
    ax = axes[idx]
    datos_año = costo_mensual[costo_mensual['Año'] == año]
    
    meses = datos_año['Mes'].values
    acumulado = datos_año['Acumulado'].values
    
    # Línea acumulada
    ax.plot(meses, acumulado, marker='o', linewidth=2.5, color='#1f77b4',
            label='Costo Acumulado', zorder=3)
    
    # Línea horizontal del presupuesto anual
    ax.axhline(y=presupuesto_anual, color='#d62728', linestyle='--', linewidth=2,
               label=f'Presupuesto Anual (${presupuesto_anual:,.0f})', zorder=4)
    
    # Detectar el mes donde se cruza el presupuesto
    cruce = None
    for i in range(len(acumulado)):
        if acumulado[i] >= presupuesto_anual:
            cruce = i
            break
    
    if cruce is not None:
        # Marcar punto de cruce
        ax.plot(meses[cruce], acumulado[cruce], 'o', markersize=12,
                color='#d62728', zorder=5, markeredgecolor='white', markeredgewidth=2)
        
        exceso_final = acumulado[-1] - presupuesto_anual
        pct = (exceso_final / presupuesto_anual) * 100
        
        # Anotación en el punto de cruce
        ax.annotate(f'Superado en {meses_labels[meses[cruce]-1]}',
                    xy=(meses[cruce], acumulado[cruce]),
                    xytext=(-60, -35), textcoords='offset points',
                    fontsize=9, fontweight='bold', color='#d62728',
                    arrowprops=dict(arrowstyle='->', color='#d62728', lw=1.5))
        
        # Sombrear zona sobre presupuesto
        mask_sobre = acumulado >= presupuesto_anual
        meses_sobre = meses[mask_sobre]
        acum_sobre = acumulado[mask_sobre]
        ax.fill_between(meses_sobre, presupuesto_anual, acum_sobre,
                        alpha=0.2, color='#e74c3c', label='Exceso')
        
        # Anotación del exceso final
        ax.annotate(f'+${exceso_final:,.0f}\n(+{pct:.1f}%)',
                    xy=(meses[-1], acumulado[-1]),
                    xytext=(10, 10), textcoords='offset points',
                    fontsize=9, fontweight='bold', color='#c0392b',
                    arrowprops=dict(arrowstyle='->', color='#c0392b', lw=1.5))
    else:
        # No se superó el presupuesto
        ahorro = presupuesto_anual - acumulado[-1]
        pct = (ahorro / presupuesto_anual) * 100
        ax.annotate(f'Bajo ppto: -${ahorro:,.0f}\n(-{pct:.1f}%)',
                    xy=(meses[-1], acumulado[-1]),
                    xytext=(10, 10), textcoords='offset points',
                    fontsize=9, fontweight='bold', color='#27ae60',
                    arrowprops=dict(arrowstyle='->', color='#27ae60', lw=1.5))
    
    ax.set_xlabel('Mes', fontsize=11)
    ax.set_ylabel('Costo Acumulado (USD)', fontsize=11)
    ax.set_title(f'Costo Acumulado vs Presupuesto Anual — {año}', fontsize=13, fontweight='bold')
    ax.set_xticks(range(1, 13))
    ax.set_xticklabels(meses_labels, fontsize=9)    ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
    ax.legend(loc='upper left', fontsize=9)
    ax.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```
