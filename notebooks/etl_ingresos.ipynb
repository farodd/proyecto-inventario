```python
import altair as alt

# Convertir fecha a datetime
ingresos_plot = ingresos.copy()
ingresos_plot['fecha'] = pd.to_datetime(ingresos_plot['FECHA DE INGRESO'], errors='coerce')
ingresos_plot = ingresos_plot.dropna(subset=['fecha', 'CLASIFICACION', 'TOTAL NETO'])

# Agrupar por mes y clasificación
ingreso_mensual = (
    ingresos_plot
    .groupby([pd.Grouper(key='fecha', freq='MS'), 'CLASIFICACION'])['TOTAL NETO']
    .sum()
    .reset_index()
)
ingreso_mensual.rename(columns={
    'TOTAL NETO': 'Costo Total',
    'CLASIFICACION': 'Categoría'
}, inplace=True)

# Gráfico interactivo con Altair
chart = alt.Chart(ingreso_mensual).mark_bar().encode(
    x=alt.X('yearmonth(fecha):T', title='Mes', axis=alt.Axis(format='%b %Y', labelAngle=-45)),
    y=alt.Y('Costo Total:Q', title='Costo Total', axis=alt.Axis(format='$,.0f')),
    color=alt.Color('Categoría:N', title='Categoría'),
    xOffset='Categoría:N',
    tooltip=[
        alt.Tooltip('yearmonth(fecha):T', title='Mes', format='%b %Y'),
        alt.Tooltip('Categoría:N'),
        alt.Tooltip('Costo Total:Q', format='$,.2f')
    ]
).properties(
    width=900,
    height=450,
    title='Ingreso Mensual por Categoría'
).interactive()

chart
```
